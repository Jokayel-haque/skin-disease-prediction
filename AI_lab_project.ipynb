```python
import tensorflow as tf
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import GlobalAveragePooling2D, Dense
from tensorflow.keras.models import Model
```

```python
# Define input shape (same as image_size used for datasets)
IMG_HEIGHT = 224
IMG_WIDTH = 224
INPUT_SHAPE = (IMG_HEIGHT, IMG_WIDTH, 3) # 3 for RGB channels
NUM_CLASSES = 2 # Based on the dataset output 'Found 7125 files belonging to 2 classes.'

# Load the MobileNetV2 model pre-trained on ImageNet
base_model = MobileNetV2(input_shape=INPUT_SHAPE,
                         include_top=False,
                         weights='imagenet')

# Freeze the base model layers to prevent re-training them
base_model.trainable = False

# Add custom classification head
x = base_model.output
x = GlobalAveragePooling2D()(x)
predictions = Dense(NUM_CLASSES, activation='softmax')(x)

# Create the full model
model = Model(inputs=base_model.input, outputs=predictions)
```

```python
# Compile the model
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

## Define Prediction Function

### Subtask:
Create a Python function that takes an image file path as input, preprocesses it to match the model's input requirements, and then uses the loaded model to predict the class.


**Reasoning**:
To define the prediction function, I need to import the required libraries, load the model, preprocess the input image according to the model's training requirements, make a prediction, and then interpret the prediction to return the class label.



```python
import tensorflow as tf
import numpy as np
from PIL import Image
import os

def predict_skin_disease(image_path):
    # Define class names based on the order found in the dataset
    class_names = ['Infected', 'Not_infected'] # Assuming this order based on the dataset

    # Load the trained model
    model_save_path = '/content/drive/MyDrive/'
    model_filename = 'mobilenetv2_skin_disease_classifier.keras'
    full_model_path = os.path.join(model_save_path, model_filename)
    loaded_model = tf.keras.models.load_model(full_model_path)

    # Load and preprocess the image
    img = tf.keras.preprocessing.image.load_img(image_path, target_size=(224, 224))
    img_array = tf.keras.preprocessing.image.img_to_array(img)
    img_array = img_array / 255.0  # Normalize pixel values
    img_array = np.expand_dims(img_array, axis=0)  # Add batch dimension

    # Make prediction
    predictions = loaded_model.predict(img_array)

    # Interpret prediction
    predicted_class_index = np.argmax(predictions[0])
    predicted_class_name = class_names[predicted_class_index]

    return predicted_class_name

print("Prediction function 'predict_skin_disease' defined.")
```

```python
import gradio as gr

# Create the Gradio interface
iface = gr.Interface(fn=predict_skin_disease,
                     inputs=gr.Image(type='filepath', label='Upload Skin Image'),
                     outputs='text',
                     title='Skin Disease Prediction',
                     description='Upload an image of skin to predict if it is Infected or Not_infected.')

# Launch the interface
iface.launch(debug=True, share=True)
print("Gradio interface launched.")
```
